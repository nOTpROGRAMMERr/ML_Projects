```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Sample documents
documents = [
    "This is the first document.",
    "Here's another document about AI.",
    "Storing vectors in a database is efficient.",
    "Nitish K Reddy is currently best batsman in Indian Test Cricket Team"
]

# Encode documents into vectors
document_vectors = model.encode(documents)

# Verify vector shape
print(f"Vector shape: {np.array(document_vectors).shape}")
```

```python
!pip install faiss-cpu
```

```python
import faiss
import numpy as np

# Dimensions of the vectors
vector_dimension = document_vectors[0].shape[0]

# Create FAISS index
index = faiss.IndexFlatL2(vector_dimension)
index.add(np.array(document_vectors))  # Add vectors to index

# Save index
faiss.write_index(index, "document_vectors.index")

# Load index
index = faiss.read_index("document_vectors.index")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

query = "who plays in Indian Test Cricket Team?"
query_vector = model.encode([query])

# Compute similarity
similarities = cosine_similarity(query_vector, document_vectors)
most_similar_idx = np.argmax(similarities)

print(f"Most similar document: {documents[most_similar_idx]}")
print(f"Similarity score: {similarities[0][most_similar_idx]}")
```
